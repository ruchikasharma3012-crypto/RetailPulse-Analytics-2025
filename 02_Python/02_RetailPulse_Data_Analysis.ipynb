```python
category_sales = (
    df.groupby("Category")["Sales_Amount"]
    .sum()
    .sort_values(ascending=False)
)

category_sales
```

```python
category_sales = (
    df.groupby("Category")["Sales_Amount"]
    .sum()
    .sort_values(ascending=False)
)

category_sales
```

```python
category_margin = (
    df.groupby("Category")["Profit_Margin_%"]
    .mean()
    .sort_values(ascending=False)
)

category_margin
```

```python
category_analysis = df.groupby("Category").agg(
    Total_Sales=("Sales_Amount", "sum"),
    Total_Profit=("Profit", "sum"),
    Avg_Profit_Margin=("Profit_Margin_%", "mean"),
    Total_Quantity=("Quantity", "sum")
).sort_values("Total_Sales", ascending=False)

category_analysis
```

```python
best_sales_category = category_analysis["Total_Sales"].idxmax()
best_profit_category = category_analysis["Total_Profit"].idxmax()
best_margin_category = category_analysis["Avg_Profit_Margin"].idxmax()

print("Highest Sales Category:", best_sales_category)
print("Highest Profit Category:", best_profit_category)
print("Highest Profit Margin Category:", best_margin_category)
```
