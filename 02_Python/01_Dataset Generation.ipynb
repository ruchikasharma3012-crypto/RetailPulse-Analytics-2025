```python
payment_modes = [
    "UPI",
    "Credit Card",
    "Debit Card",
    "Cash",
    "Net Banking"
]

customer_segments = [
    "Regular",
    "Premium",
    "Corporate"
]

sales_channels = [
    "Online",
    "Offline"
]

salespersons = [
    "Amit Sharma",
    "Priya Verma",
    "Rahul Mehta",
    "Neha Singh",
    "Rohit Gupta",
    "Anjali Kapoor",
    "Vikas Yadav",
    "Sneha Iyer"
]
```

```python
sales_data = []
```

```python
for i in range(1, 10001):

    order_id = f"ORD{i:06d}"

    customer_name = fake.name()

    gender = random.choice(["Male", "Female"])

    age = random.randint(18, 65)

    city = random.choice(cities)
    state = state_map[city]

    product = random.choice(list(product_master.keys()))

    details = product_master[product]

    category = details["Category"]
    brand = details["Brand"]
    unit_price = details["Unit_Price"]
    cost_price = details["Cost_Price"]

    quantity = random.randint(1, 5)

    discount_percent = random.choice([0, 5, 10, 15, 20])

    sales_amount = quantity * unit_price

    discount_amount = round(
        sales_amount * discount_percent / 100,
        2
    )

    taxable_amount = sales_amount - discount_amount

    gst = round(taxable_amount * 0.18, 2)

    final_amount = round(
        taxable_amount + gst,
        2
    )

    total_cost = quantity * cost_price

    profit = round(
        taxable_amount - total_cost,
        2
    )

    profit_margin = round(
        (profit / taxable_amount) ,
        4
    ) if taxable_amount > 0 else 0

    payment_mode = random.choice(payment_modes)

    customer_segment = random.choice(customer_segments)

    sales_channel = random.choice(sales_channels)

    salesperson = random.choice(salespersons)

    order_date = fake.date_between(
        start_date="-2y",
        end_date="today"
    )

    sales_data.append([
        order_id,
        customer_name,
        gender,
        age,
        city,
        state,
        category,
        product,
        brand,
        quantity,
        unit_price,
        cost_price,
        total_cost,
        discount_percent,
        discount_amount,
        sales_amount,
        gst,
        final_amount,
        profit,
        profit_margin,
        payment_mode,
        customer_segment,
        sales_channel,
        salesperson,
        order_date
    ])
```

```python
columns = [
    "Order_ID",
    "Customer_Name",
    "Gender",
    "Age",
    "City",
    "State",
    "Category",
    "Product",
    "Brand",
    "Quantity",
    "Unit_Price",
    "Cost_Price",
    "Total_Cost",
    "Discount_%",
    "Discount_Amount",
    "Sales_Amount",
    "GST",
    "Final_Amount",
    "Profit",
    "Profit_Margin_%",
    "Payment_Mode",
    "Customer_Segment",
    "Sales_Channel",
    "Salesperson",
    "Order_Date"
]

df = pd.DataFrame(sales_data, columns=columns)
```

```python
df["Order_Date"] = pd.to_datetime(df["Order_Date"])

df["Order_Month"] = df["Order_Date"].dt.month_name()

df["Quarter"] = "Q" + df["Order_Date"].dt.quarter.astype(str)

df["Day_Name"] = df["Order_Date"].dt.day_name()

region_map = {
    "Delhi": "North",
    "Mumbai": "West",
    "Bengaluru": "South",
    "Hyderabad": "South",
    "Chennai": "South",
    "Kolkata": "East",
    "Pune": "West",
    "Ahmedabad": "West",
    "Jaipur": "North",
    "Lucknow": "North"
}

df["Region"] = df["City"].map(region_map)


df["Delivery_Days"] = np.random.randint(1, 8, len(df))


df["Customer_Rating"] = np.random.choice(
    [3, 4, 5],
    size=len(df),
    p=[0.15, 0.35, 0.50]
)


df["Return_Status"] = np.random.choice(
    ["Returned", "Not Returned"],
    size=len(df),
    p=[0.08, 0.92]
)


df["Coupon_Used"] = np.random.choice(
    ["Yes", "No"],
    size=len(df),
    p=[0.35, 0.65]
)


df["Loyalty_Member"] = np.random.choice(
    ["Yes", "No"],
    size=len(df),
    p=[0.40, 0.60]
)


df["Customer_Type"] = np.random.choice(
    ["New", "Returning"],
    size=len(df),
    p=[0.35, 0.65]
)


df["Payment_Status"] = np.random.choice(
    ["Paid", "Pending"],
    size=len(df),
    p=[0.95, 0.05]
)


df["Delivery_Status"] = np.random.choice(
    ["Delivered", "In Transit", "Cancelled"],
    size=len(df),
    p=[0.88, 0.10, 0.02]
)


df["Stock_Status"] = np.random.choice(
    ["In Stock", "Low Stock", "Out of Stock"],
    size=len(df),
    p=[0.70, 0.20, 0.10]
)


def get_season(month):
    if month in [12, 1, 2]:
        return "Winter"
    elif month in [3, 4, 5]:
        return "Summer"
    elif month in [6, 7, 8, 9]:
        return "Monsoon"
    else:
        return "Festive"

df["Season"] = df["Order_Date"].dt.month.apply(get_season)
```
